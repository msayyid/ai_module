### (a) Formulate the robot-maze problem (the simple “room with an exit gap”)

**State (what you must store):**
A natural state representation is:

* **(x, y, dir)** where

  * (x, y) = robot position
  * dir ∈ {N, E, S, W} = robot’s facing direction

**Initial state:**

* Robot starts **in the center** of the room, facing **North**

  * s0 = (x_center, y_center, N)

**Goal test:**

* “Robot is at the exit / outside the room.”
  Examples:

  * robot’s position crosses the exit opening, or
  * robot reaches a special “EXIT” cell/region.

**Actions:**
You have two types (as described):

1. **TurnTo(N/E/S/W)** (set orientation)
2. **MoveForward(d)** (move forward some distance; robot stops before the wall if d is too large)

**Transition model (what actions do):**

* TurnTo changes only `dir`
* MoveForward changes (x, y) by moving along `dir` until either:

  * distance d is covered, or
  * the robot would hit a wall, so it stops at the last valid position.

**Cost function (typical choices):**

* Turning cost: 1 per turn (or 0 if you do not care about turning)
* Moving cost: distance moved (or 1 per move if you want “number of moves”)
  So one common choice:
* **cost = (distance moved) + (turn penalty)**

**How large is the state space?**

* If (x, y) is **continuous** (any real coordinate in the room), then the state space is **infinite (uncountably infinite)**.
* In most AI search assignments we **discretize** the room into a grid of free cells:

  * If there are **K** free positions/cells inside the room, then
    **|S| = 4K** (because 4 possible orientations for each position).

### (b) Reformulate using the “only turn at intersections” observation

Now we compress the maze into a **graph of decision points**.

**Key idea:**
You do not need states for every corridor cell. You only care about:

* intersections (degree ≥ 3, or “2 or more corridors” as your prompt says),
* dead ends (optional but often included),
* the start,
* the exit.

**State:**

* **current node = which intersection (or start/exit) you are at**
  Optionally include the direction you arrived from (useful to prevent immediate backtracking):
* (node, incoming_edge) or (node, incoming_dir)

**Initial state:**

* node = StartIntersection (the junction closest to the robot’s start location, or treat the start as its own node)

**Goal test:**

* node == ExitNode

**Actions:**

* At an intersection, choose which corridor to take next:

  * **GoTo(neighbor_node)** for each corridor leading out

**Transition model:**

* Taking a corridor moves you directly to the next intersection (or exit/dead end).

**Cost function:**

* Cost of an action = **length of the corridor** (number of steps/cells, or geometric distance).
  This makes shortest-path search (like UCS / A*) very natural.

**Effect:**
State space becomes much smaller: instead of “all corridor cells × 4 directions”, it is roughly:

* **number of intersections (and special nodes)** (and maybe a small factor if you track incoming direction).
