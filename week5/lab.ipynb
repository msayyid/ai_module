```python
def a_star_algo(start_node, stop_node):
    open_set = {start_node} # nodes discovered but not fully expanded yet (frontier)
    closed_set = set() # it stores nodes already expanded/processed
    g = {} # best known cost from start to each node

    # For each node, which node we came from to reach it with the best path found so far.
    parents = {} # best predecessor for each node (used to reconstruct the final path)

    # distance of starting node from itself is zero
    g[start_node] = 0
    # start_node is a root node, has no parents (can be considered a parent to itself)
    # start_node is set to its own parent node

    parents[start_node] = start_node

    while(len(open_set)):
        n = None
        # node with the lowest f() is found
        for v in open_set:
            # Choose the node in open_set with the lowest f = g + h.
            if n == None or g[v] + get_heuristic(v) < g[n] + get_heuristic(n):
                n = v
# ------------------------------------------------------------------------------------
        # if n == stop_node or graph_nodes[n] == []:
        #     # if n is the last node or if n has no neighbors: do nothing/don't expand neighbors
        #     pass
        # else:
        for (m, weight) in get_neighbors(n): 
            if m not in open_set and m not in closed_set:
                open_set.add(m) # new node to consider later
                parents[m] = n # record how we got to m
                g[m] = g[n] + weight # update the actual cost to reach m from n
            
            else:
                # !!!!!!!!!
                if g[m] > g[n] + weight:
                    # update g(m)
                    g[m] = g[n] + weight
                    parents[m] = n

                    if m in closed_set:
                        closed_set.remove(m)
                        open_set.add(m)
        if n == None:
            print("Path does not exist")
            return None

        # if the current node is the stop_node
        # then we begin reconstructing the path from it to the start node
        if n == stop_node:
            path = []

            while parents[n] != n: # loop until you reach the starting node
                path.append(n)        
                n = parents[n]

            path.append(start_node)
            path.reverse()
            print("Path found: {}".format(path))
            return path
        # remove n from the open list, and add it to the closed list
        # because all of its neighbors were inspected
        open_set.remove(n)
        closed_set.add(n)
    print("Path does not exist")
    return None


def get_heuristic(node):
    return heuristics[node]

def get_neighbors(node):
    return (graph_nodes[node])



if __name__ == "__main__":
    # graph_nodes = {
        # "A": [("B", 2), ("E", 3)],
        # "B": [("C", 1), ("G", 9)],
        # "C": [],
        # "E": [("D", 6)],
        # "D": [("G", 1)],
        # "G": []
    # }
    # heuristics = {
    #     "A": 11,
    #     "B": 6,
    #     "C": 99,
    #     "G": 0,
    #     "E": 7,
    #     "D": 1
    # }
    # a_star_algo("A", "G")

    graph_nodes = {
        "A": [("B", 2), ("D", 3)],
        "B": [("A", 2), ("C", 4), ("E", 1)],
        "C": [("B", 4), ("F", 2)],
        "D": [("A", 3), ("E", 2)],
        "E": [("D", 2), ("B", 1), ("F", 1)],
        "F": [("E", 1), ("C", 2), ("G", 3)],
        "G": [("F", 3)]
    }

    heuristics = {
        "A": 6,
        "B": 4,
        "C": 2,
        "D": 5,
        "E": 3,
        "F": 1,
        "G": 0
    }

    a_star_algo("A", "G")
```
