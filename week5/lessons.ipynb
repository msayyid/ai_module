# Uninformed Search

## 1. Uninformed Search
Uninformed search means:
The algorithm has no extra knowledge about how close a state is to the goal.
It only knows:
Initial state
Successor function
Goal test
Path cost (sometimes)
No heuristic. No guidance.
It blindly explores.


## 2. Depth-First Search (DFS)
### Strategy
Expand the deepest node first.
Think:
Go down one branch as far as possible, then backtrack.
Implementation:
Stack
Or recursion
### What nodes does DFS expand?
Some left prefix of the tree.
In worst case, it may explore the entire tree.
### Is DFS complete?
Only if:
The search space is finite
Or you prevent cycles
If the tree is infinite and you don’t prevent revisiting nodes, it can loop forever.
So:
DFS is not complete in general.
### Is DFS optimal?
No.
It returns the first solution it finds.
That might:
Be very deep
Have high cost
DFS does not care about:
Depth
Cost
It just goes leftmost and deep.

### Complexity
Let:
b = branching factor
m = maximum depth
#### Time complexity:
O(b^m)
#### Space complexity:
O(bm)
#### Important:
DFS uses much less memory than BFS.

***** 
### What is b (branching factor)?
The branching factor (b) is:
The average number of children (successors) each node has.
In simple terms:
If from every state you can go to 3 new states,
then b = 3.

## 3. Breadth-First Search (BFS)
Strategy
Expand the shallowest node first.

Think:
Explore level by level.

Implementation:
Queue

### What nodes does BFS expand?

If the shallowest solution is at depth d:

It expands:

All nodes at depth 0

All nodes at depth 1

...

All nodes at depth d

### Is BFS complete?

Yes, if:

Branching factor is finite

Because:
If a solution exists at depth d,
BFS will eventually reach that level.

### Is BFS optimal?

Only if:

All step costs are equal (usually 1)

Because:
It finds the shallowest solution.
If each step costs the same,
shallowest = cheapest.

If costs vary,
BFS is not optimal.

### Complexity

Let:

b = branching factor

d = depth of shallowest solution

d = number of moves from start to the first goal BFS would find.

#### Time:
O(b^d)

#### Space:
O(b^d)

Important:
BFS uses a lot of memory.

## 4. Uniform Cost Search (UCS)
Strategy

Expand the cheapest node first (lowest path cost g(n)).

Implementation:

Priority queue (min-heap)

This is like:
#### BFS but ordered by cost instead of depth.

### What nodes does UCS expand?

It expands all nodes with cost less than the optimal solution cost C*.

If:

Minimum action cost = e

Optimal solution cost = C*

Then effective depth ≈ C* / e

### Is UCS complete?

Yes, if:

Step costs are positive

Optimal solution cost is finite

Because cost keeps increasing,
so it will eventually reach the solution.

### Is UCS optimal?

Yes.

It always returns the lowest cost solution.

This is very important.

### Complexity

Time:
O(b^(C*/e))

Space:
Also exponential

Like BFS, it uses a lot of memory.

****
#### C* - C star: The cost of the optimal (cheapest) solution; This is a total path cost
Path 1:
S → A → G
Cost = 2 + 2 = 4

Path 2:
S → B → G
Cost = 5 + 5 = 10

The cheapest path costs 4.

So:

C* = 4

****
#### e = minimum step cost in the whole problem

In the example above:

Edges cost either 2 or 5
Smallest = 2

So:

e = 2

### Why Time Complexity is O(b^(C*/e))

Now we connect it to BFS.

For BFS:

Time = O(b^d)

where d = solution depth.

For UCS:

Depth is not fixed.
Instead, we estimate worst-case depth using:

d ≈ C* / e

So we replace d with C*/e:

Time ≈ O(b^(C*/e))

### Big Intuition Summary

C* = total cost of best solution
e = smallest single-step cost

C*/e ≈ maximum number of steps UCS might explore

So:

Time ≈ O(b^(C*/e))


## Search Algorithm Comparison

**Definitions**

- `b` = branching factor  
- `d` = depth of shallowest solution  
- `m` = maximum depth of the tree  
- `C*` = optimal solution cost  
- `e` = minimum step cost  

---

| Algorithm | Complete? | Optimal? | Time Complexity | Space Complexity |
|------------|------------|------------|------------------|------------------|
| **DFS** | No (unless finite and no cycles) | No | O(b^m) | O(bm) |
| **BFS** | Yes (if b is finite) | Only if all step costs = 1 | O(b^d) | O(b^d) |
| **UCS** | Yes (if step costs > 0) | Yes | O(b^(C*/e)) | O(b^(C*/e)) |

---

### Key Intuition

- **DFS** → Go deep first  
- **BFS** → Search level by level  
- **UCS** → Expand cheapest path first  


# Informed Search

## 1️. What is Informed Search?

Uninformed search:

No extra knowledge.

Just blindly explores.

Informed search:

Uses problem-specific knowledge.

Uses a function to estimate how close we are to the goal.

That estimate is called a:

Heuristic function → h(n)

## 2. Heuristic Function h(n)

Definition:

h(n) = estimated cost from node n to the goal

Important:

It is an estimate.

It does NOT know the true remaining cost.

It should never be negative.

If n is the goal → h(n) = 0.


### Romania Example

Goal = Bucharest.

h(n) = straight-line distance to Bucharest.

Example:

Arad → 366

Sibiu → 253

Bucharest → 0

This is intuitive:

The closer a city is (in straight-line distance),
the smaller h(n) is.

## 3. Best-First Search

In informed search, we expand the node with smallest:

f(n)

Different algorithms choose different f(n).

## 4. Greedy Best-First Search

Greedy uses:

f(n) = h(n)

That means:
It ignores the cost so far.
It only looks at:
"Which node looks closest to the goal?"

### Example: From Arad to Bucharest

Start at Arad.

Neighbors:

Sibiu (253)

Timisoara (329)

Zerind (374)

Smallest h(n) = 253 → choose Sibiu.

Then from Sibiu:

Look at its neighbors.

Pick the one with smallest h(n).

Eventually:
It reaches Bucharest.


### Is Greedy Optimal?

No.

Because it ignores g(n) (the cost already spent).

It may choose a path that:
Looks close,
But is actually expensive.

Greedy is fast, but risky.

## 5. A* Search

Now the important one.

A* uses:

f(n) = g(n) + h(n)

Where:

g(n) = cost from start to n

h(n) = estimated cost from n to goal

So:

f(n) = estimated total cost of solution through n

This is very important:

A* balances:

What we already spent (g)

What we estimate remains (h)


### Why A* is Powerful

If h(n) is good (admissible and consistent),
then:

A* is complete

A* is optimal

It will find the cheapest path.

## 6. Compare the Three

UCS:
f(n) = g(n)

Greedy:
f(n) = h(n)

A*:
f(n) = g(n) + h(n)

So:

UCS = careful but blind
Greedy = optimistic but risky
A* = balanced and intelligent

## Comparison: UCS vs Greedy vs A*

| Algorithm | f(n) | What does it care about? | Uses past cost g(n)? | Uses heuristic h(n)? | Optimal? |
|------------|--------|--------------------------|----------------------|----------------------|----------|
| UCS | g(n) | Cheapest path so far | Yes | No | Yes |
| Greedy | h(n) | Looks closest to goal | No | Yes | No |
| A* | g(n) + h(n) | Best total estimated cost | Yes | Yes | Yes (if h is admissible) |


### 🔵 Uniform Cost Search (UCS)

UCS chooses the node with smallest:

g(n) = cost from start to that node

It only cares about what you have already spent.

It does NOT care how close you are to goal.

It says:

#### “So far, which path has cost me the least?”

It behaves like Dijkstra’s algorithm.


### 🔴 Greedy Search

Greedy chooses the node with smallest:

h(n) = estimated distance to goal

It does NOT care how much it already spent.

It says:

#### “Which node looks closest to the goal?”

It ignores the cost so far.


### VERY SIMPLE NUMERIC EXAMPLE

Imagine this:

You want to reach the goal.

You have two choices:

Path A:

You already spent 100

Heuristic says 1 more

Path B:

You already spent 10

Heuristic says 50 more

So:

Node A:

g = 100

h = 1


Node B:

g = 10

h = 50


Now:

#### UCS looks only at g:

Choose B (10 < 100)

#### Greedy looks only at h:

Choose A (1 < 50)

A* looks at g + h:

Node A → 100 + 1 = 101
Node B → 10 + 50 = 60

Choose B

## 7. Romania A* Example (Conceptually)

From Arad:

Arad:
g = 0
h = 366
f = 366

Expand Arad.

Sibiu:
g = 140
h = 253
f = 393

Timisoara:
g = 118
h = 329
f = 447

Zerind:
g = 75
h = 374
f = 449

Smallest f = 393 → choose Sibiu.

This continues until Bucharest is chosen.

Eventually A* finds the optimal path:
Arad → Sibiu → Rimnicu Vilcea → Pitesti → Bucharest

Total cost = 418

### The One Sentence That Fixes Everything

g(n) = real cost from start to n
h(n) = estimated cost from n to goal

We add them to estimate total trip cost.

```python
import heapq
from collections import deque

# --- Data Setup ---

# The graph represents connections and their actual weights (g_cost)
graph = {
    "A": {"B": 1, "C": 4},
    "B": {"D": 2, "E": 5},
    "C": {"F": 3},
    "D": {"G": 1},
    "E": {"G": 2},
    "F": {"G": 5},
    "G": {}
}

# Heuristic (h_cost): Estimated distance from current node to "G"
heuristic = {
    "A": 7, "B": 6, "C": 4, "D": 2, "E": 1, "F": 3, "G": 0
}

start_node = "A"
goal_node = "G"

def print_header(name):
    print(f"\n{'='*10} Running {name} {'='*10}")

# --- 1. Depth First Search (DFS) ---
def dfs(start, goal):
    print_header("DFS (Last-In, First-Out)")
    # Stack stores (current_node, path_taken)
    stack = [(start, [start])]
    visited = set()

    while stack:
        node, path = stack.pop()
        print(f"Visiting: {node} | Path so far: {' -> '.join(path)}")

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)
            # We iterate in reverse to maintain alphabetical order in the stack
            for neighbor in reversed(list(graph[node].keys())):
                if neighbor not in visited:
                    stack.append((neighbor, path + [neighbor]))
    return None

# --- 2. Breadth First Search (BFS) ---
def bfs(start, goal):
    print_header("BFS (First-In, First-Out)")
    # Queue stores (current_node, path_taken)
    queue = deque([(start, [start])])
    visited = set()

    while queue:
        node, path = queue.popleft()
        print(f"Visiting: {node} | Path so far: {' -> '.join(path)}")

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append((neighbor, path + [neighbor]))
    return None

# --- 3. Uniform Cost Search (UCS) ---
def ucs(start, goal):
    print_header("UCS (Cheapest Actual Path First)")
    # Priority Queue: (cumulative_g_cost, current_node, path)
    pq = [(0, start, [start])]
    visited = {} # Node: cheapest_cost_to_reach

    while pq:
        cost, node, path = heapq.heappop(pq)
        print(f"Expanding: {node} | Cumulative Cost: {cost} | Path: {' -> '.join(path)}")

        if node == goal:
            return path, cost

        # Only explore if we found a shorter way to this node than before
        if node not in visited or cost < visited[node]:
            visited[node] = cost
            for neighbor, weight in graph[node].items():
                new_cost = cost + weight
                heapq.heappush(pq, (new_cost, neighbor, path + [neighbor]))
    return None

# --- 4. Greedy Best-First Search ---
def greedy(start, goal):
    print_header("Greedy (Closest by Heuristic Only)")
    # Priority Queue: (h_cost, current_node, path)
    pq = [(heuristic[start], start, [start])]
    visited = set()

    while pq:
        h, node, path = heapq.heappop(pq)
        print(f"Expanding: {node} | Heuristic (h): {h} | Path: {' -> '.join(path)}")

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)
            for neighbor in graph[node]:
                heapq.heappush(pq, (heuristic[neighbor], neighbor, path + [neighbor]))
    return None

# --- 5. A* Search ---
def astar(start, goal):
    print_header("A* (Actual Cost + Heuristic)")
    # Priority Queue: (f_cost, g_cost, current_node, path)
    # f = g (cost so far) + h (estimated cost to goal)
    pq = [(heuristic[start], 0, start, [start])]
    visited = {}

    while pq:
        f, g, node, path = heapq.heappop(pq)
        print(f"Expanding: {node} | f={f} (g:{g} + h:{heuristic[node]}) | Path: {' -> '.join(path)}")

        if node == goal:
            return path, g

        if node not in visited or g < visited[node]:
            visited[node] = g
            for neighbor, weight in graph[node].items():
                new_g = g + weight
                new_f = new_g + heuristic[neighbor]
                heapq.heappush(pq, (new_f, new_g, neighbor, path + [neighbor]))
    return None

# --- Execution ---

results = {
    "DFS": dfs(start_node, goal_node),
    "BFS": bfs(start_node, goal_node),
    "UCS": ucs(start_node, goal_node),
    "Greedy": greedy(start_node, goal_node),
    "A*": astar(start_node, goal_node)
}

print("\n" + "#" * 30)
print("FINAL RESULTS SUMMARY")
print("#" * 30)
for alg, res in results.items():
    print(f"{alg:8}: {res}")
```
